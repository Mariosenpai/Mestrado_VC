```python
from transformers import pipeline

translator = pipeline("text-generation", model="liaad/Tradutor")
```

```python
import re

# Função auxiliar para dividir o texto em blocos menores sem cortar no meio das frases
def dividir_em_blocos(texto, max_chars=2500):
    """
    Divide o texto em blocos menores que 'max_chars', 
    preferindo cortar em pontos finais, exclamações ou interrogações.
    """
    blocos = []
    
    for p in texto.split("\n\n"):
    
        if len(p) > max_chars:
        
            while len(texto) > max_chars:
                # Procura o último ponto antes do limite
                corte = texto.rfind('.', 0, max_chars)
                if corte == -1:
                    # Se não houver ponto, tenta vírgula ou espaço
                    corte = texto.rfind(',', 0, max_chars)
                if corte == -1:
                    corte = texto.rfind(' ', 0, max_chars)
                if corte == -1:
                    # Se não achar ponto natural, corta direto (último recurso)
                    corte = max_chars
        
                blocos.append(texto[:corte+1].strip())
                texto = texto[corte+1:].strip()
            if texto:
                blocos.append(texto.strip())    
        else:
            blocos.append(p.strip())
    
    return blocos
```

```python
from pathlib import Path


pasta = Path("outputs")


for p in pasta.iterdir():
    if p.is_file():
        print(f"\n=== {p.name} ===")
        vecto_text = []

        # Lê o conteúdo do arquivo
        conteudo = p.read_text(encoding="utf-8")

        # Divide o texto por blocos/parágrafos (linhas vazias)
        paragrafos = dividir_em_blocos(conteudo )#[p.strip() for p in conteudo.split("\n\n") if p.strip()]

        # Mostra cada parágrafo separado
        for i, paragrafo in enumerate(paragrafos, start=1):
            print(f"\n[Parágrafo {i}]\n{paragrafo}")

            chat = [
                    {
                        "role": "system",
                        "content": "You are a translator from English to Brazilian Portuguese",
                },
                {
                    "role": "user",
                    "content": f"Translate this text from English to Brazilian Portuguese: {paragrafo}",
                },
            ]

            

            translated_text = translator(
                chat, 
                max_length=1024,
                pad_token_id=translator.model.config.eos_token_id  # Not necessary. Just to avoid warning.
            )
            text_traduzido = translated_text[-1]["generated_text"][-1]["content"]
            print(f"\n{text_traduzido}\n{'-'*50}")
            vecto_text.append(text_traduzido)
            
            
        vecto_text = " \n".join(vecto_text)
        open(f"traduzido/{p.name}.md", "w", encoding="utf-8").write(vecto_text)
        print("----------------- Traduzido!! -----------------")
```
