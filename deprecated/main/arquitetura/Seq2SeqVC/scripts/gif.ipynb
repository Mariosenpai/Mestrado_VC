```python
import os
import imageio
import re
pasta_raiz = r"C:\Users\USER\Documents\Mestrado\codigo\Mestrado_VC\main\arquitetura\Seq2SeqVC\exp\aas_vc_teste_encoder_dp\predictions"

# ordenar pastas de epoca
epocas = sorted(
    [p for p in os.listdir(pasta_raiz)
     if os.path.isdir(os.path.join(pasta_raiz, p))],
    key=lambda x: int(x.replace("steps", ""))
)

# pegar imagens da primeira epoca dentro de outs
primeira_epoca = os.path.join(pasta_raiz, epocas[0], "outs")
imagens = [f for f in os.listdir(primeira_epoca) if f.endswith(".png")]

for img_nome in imagens:

    frames = []

    for epoca in epocas:

        caminho = os.path.join(pasta_raiz, epoca, "outs", img_nome)

        if os.path.exists(caminho):
            frames.append(imageio.imread(caminho))

    if frames:
        gif_nome = img_nome.replace(".png", ".gif")
        imageio.mimsave(gif_nome, frames, duration=0.5)

        print(f"GIF criado: {gif_nome}")
```
